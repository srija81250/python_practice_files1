```python
df.dropna(inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['OrderDate']=pd.to_datetime(df['OrderDate'])
```

```python
df['Expected_revenue']=df['Quantity']*df['TotalAmount']
```

```python
import numpy as np
df['Category']=np.where(df['Expected_revenue']>1000,'high','low')
```

```python
sales=df.groupby(['Category'])['Expected_revenue'].sum()
sales.plot(kind='bar')
plt.xlabel("Category")
plt.ylabel("Expected_revenue")
plt.show()
```

```python
df['Month_Num'] = df['OrderDate'].dt.month
df['Month_Name'] = df['OrderDate'].dt.month_name()
```

```python
monthly_sales = df.groupby(
    ['Month_Num', 'Month_Name']
)['Expected_revenue'].sum().reset_index()
```

```python
monthly_sales = monthly_sales.sort_values('Month_Num')
monthly_sales
```

```python
monthly_sales
```

```python
from matplotlib import figure
plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales['Month_Name'],
    monthly_sales['Expected_revenue'],
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.show()
```
